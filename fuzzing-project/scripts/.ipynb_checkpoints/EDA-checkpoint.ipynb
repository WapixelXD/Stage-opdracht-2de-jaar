```python
quality_counts = df['runid'].value_counts()

plt.figure(figsize=(8, 6))
plt.bar(quality_counts.index, quality_counts, color='deeppink')
plt.title('Count Plot of RunID')
plt.xlabel('RunID')
plt.ylabel('Count')
plt.show()
```

```python
sns.set_style("darkgrid")

numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns

plt.figure(figsize=(14, len(numerical_columns) * 3))
for idx, feature in enumerate(numerical_columns, 1):
    plt.subplot(len(numerical_columns), 2, idx)
    sns.histplot(df[feature], kde=True)
    plt.title(f"{feature} | Skewness: {round(df[feature].skew(), 2)}")

plt.tight_layout()
plt.show()
```

```python
sns.set_palette("Pastel1")

plt.figure(figsize=(10, 6))

sns.pairplot(df)

plt.suptitle('Pair Plot for DataFrame')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Groepeer per run en per minuut (60 sec)
df['minute_bucket'] = (df['seconds'] // 60).astype(int)
summary = df.groupby(['runid', 'minute_bucket', 'status']).size().unstack(fill_value=0)

# 2. Plot voor elke run een aparte subplot (onder elkaar)
runids = df['runid'].unique()
fig, axes = plt.subplots(len(runids), 1, figsize=(15, 4 * len(runids)), sharex=True)

color_map = {
    200: "#2ecc71", 201: "#27ae60", 400: "#f1c40f", 
    401: "#e67e22", 404: "#e74c3c", 429: "#9b59b6", 500: "#c0392b"
}

for i, rid in enumerate(runids):
    run_data = summary.loc[rid]
    run_data.plot(kind='bar', stacked=True, ax=axes[i], 
                  color=[color_map.get(col, "#bdc3c7") for col in run_data.columns],
                  width=1.0, edgecolor='none')
    
    axes[i].set_title(f"Run {rid}: Status verdeling per minuut")
    axes[i].set_ylabel("Aantal requests")
    axes[i].legend(title="Status", loc='upper right')

plt.xlabel("Tijd (minuten)")
plt.tight_layout()
plt.show()
```
